```python
import pandas as pd
import plotly.express as px
import json

# Load the dataset
df = pd.read_csv("Red Light Cameras Data - 4326.csv")

# extract coordinates
df['geometry'] = df['geometry'].apply(json.loads)
df['Longitude'] = df['geometry'].apply(lambda g: g['coordinates'][0][0])
df['Latitude'] = df['geometry'].apply(lambda g: g['coordinates'][0][1])

# Combine streets to create intersection label
df['Intersection'] = df['CLIENT_STREET_1'].fillna('') + ' & ' + df['CLIENT_STREET_2'].fillna('')

# Clean and extract year
df['ACTIVATION_DATE'] = pd.to_datetime(df['ACTIVATION_DATE'], errors='coerce')
df['Year Installed'] = df['ACTIVATION_DATE'].dt.year

# Filter valid rows
df_clean = df.dropna(subset=['Latitude', 'Longitude', 'Year Installed'])

# Plot map 
fig = px.scatter_mapbox(
    df_clean,
    lat='Latitude',
    lon='Longitude',
    hover_name='Intersection',
    hover_data={'Year Installed': True, 'Longitude': False, 'Latitude': False},
    color='Year Installed',
    color_continuous_scale='Turbo',
    zoom=10,
    height=550,
    mapbox_style='carto-positron',
    title='Red Light Camera Locations in Toronto by Installation Year'
)

fig.show()

```
