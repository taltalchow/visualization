# Visualization 1: Map of Active Red Light Cameras in Toronto (Python)

 > What software did you use to create your data visualization?
 I used Python with Plotly Express to create this map

> Who is your intended audience? 
This visualization is intended for Toronto residents, urban planners, city officials and anyone who want to understand the geographic distribution of active red light cameras in Toronto.
    
> What information or message are you trying to convey with your visualization? 
The purpose of this map is to visualize where active red light cameras are currently installed across Toronto and it will help which areas have heavier enforcement.
    
> What aspects of design did you consider when making your visualization? How did you apply them? With what elements of your plots? 
I used a simple background map so the points stand out. The color shows the year each camera was installed, using a bright color scale to make differences clear. I only showed the intersection and year to keep the map clean and easy to understand.
    
> How did you ensure that your data visualizations are reproducible? If the tool you used to make your data visualization is not reproducible, how will this impact your data visualization? 
I used Python and public data from Toronto Open Data. All the code is included in my .ipynb file, so anyone can run it again and get the same result.
    
> How did you ensure that your data visualization is accessible?  
I used clear colors and tooltips with short info. The map is interactive, so users can zoom and hover to explore. The design avoids clutter and uses good contrast.

    
> Who are the individuals and communities who might be impacted by your visualization?  
People living in areas with more cameras, city planners, journalists, and community groups might use this to understand where cameras are placed and how it's changed over time.

    
> How did you choose which features of your chosen dataset to include or exclude from your visualization? 
I included intersections and year installed because that tells the main story. I left out IDs, extra columns, and raw coordinates to keep it simple.
    
> What ‘underwater labour’ contributed to your final data visualization product?
I cleaned the data, got the year from the date, and pulled coordinates from the geometry field. I also tested different colors and tooltip settings to make the map look good and easy to use.
